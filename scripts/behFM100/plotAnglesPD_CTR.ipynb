## Here we want to explore angles and the relationship between Beh and FM100 data.

```python
from metamers.behFM100.combine import combine_behavioral_FM100

combined_df = combine_behavioral_FM100()
combined_PDCTR = combined_df[combined_df["Group"].isin(["CTR", "PD"])]

combined_PDCTR.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

def plot1variable(data = combined_PDCTR, var= "pca_angle_deg", title=""):
    data = data.copy()
    data["GroupPlot"] = data["Group"]

    # Replace CVD with its Subgroup (Protan/Deutan)
    mask = data["Group"] == "CVD"
    data.loc[mask, "GroupPlot"] = data.loc[mask, "Subgroup"]

    plt.figure(figsize=(7, 6))

    sns.boxplot(data=data, x="GroupPlot", y=var, hue="GroupPlot")

    plt.ylabel(var)
    plt.title(title)
    plt.tight_layout()
    plt.show()


plot1variable(data = combined_PDCTR, var = "pca_angle_deg", title = "Behavioral angles per group")
plot1variable(data = combined_PDCTR, var = "VKS_Angle", title = "FM100 angles per group")



```

```python
plot1variable(data = combined_PDCTR, var = "TES", title = "FM100 TES")
plot1variable(data = combined_PDCTR, var = "PES_RG", title = "FM100 PES RG")
plot1variable(data = combined_PDCTR, var = "PES_BY", title = "FM100 PES BY")
plot1variable(data = combined_PDCTR, var = "VKS_Cindex", title = "C-index")
plot1variable(data = combined_PDCTR, var = "VKS_Sindex", title = "S-index")


```

```python
plot1variable(data = combined_PDCTR, var = "cov_rg", title = "Beh cov_rg")
plot1variable(data = combined_PDCTR, var = "ellipse_area", title = "Beh ellipse area")
plot1variable(data = combined_PDCTR, var = "pc1_var", title = "var first dimension")
plot1variable(data = combined_PDCTR, var = "pc2_var", title = "var second dimension")
plot1variable(data = combined_PDCTR, var = "hull_area", title = "Beh hull area")
plot1variable(data = combined_PDCTR, var = "skew_r", title = "Beh skew red")
plot1variable(data = combined_PDCTR, var = "skew_g", title = "Beh skew green")
plot1variable(data = combined_PDCTR, var = "kurt_r", title = "Beh kurt red")
plot1variable(data = combined_PDCTR, var = "kurt_g", title = "Beh kurt green")
plot1variable(data = combined_PDCTR, var = "entropy", title = "Beh entropy")
```
